The cell above imports the package `megcourselib`. This is a folder in the course repository (have a look in the file tree of your editor, or on the left side of this page). During the setup, it was installed to be 'editable'. This means that unlike for regularly installed python packages, you can make changes to this package that become available without having to reinstall the package. 

By default, your changes will become available the next time you import the package. Unfortunately, python does not reload packages that have already been imported by default. This means that for interactive sessions, like this jupyter notebook you would have to restart the kernel for the changes to become visible.
Fortunately, jupyter notebooks have a 'line magic' to circumvent this. After running the next cell, you should be able to re-import your changes on the fly, without restarting the kernel.
